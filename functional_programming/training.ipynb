#### Exercise:

You are a teacher and have student records containing id of a stuent and the mars list in each subject where different stuents have taken different number of subjects.All nthese records are in hard copy. You want to enter all the ddata in computer and want to compute the average mars of each stuent and display.

```python
def inputStudentInfo():
    d = {}
    while True:
        firstName = input('Enter your first name')
        #subjects = input('Enter the subjects')
        scores = input('Enter the scores serially same as you entered the subjects, ' \
        'separate with coma')
        moreRecords = input('Enter "no" if no more record')
        if firstName in d:
            print('Record already exist')
        else:
            d[firstName] = scores.split(',')
        if moreRecords == 'no':
            return d

def getAvg(d):
    avgScore = {}
    for i in d:
        l = d[i]
        s = 0
        for x in l:
            s += int(x)
        avgScore[i] = s/len(l)
    return avgScore        

d = inputStudentInfo()
x = getAvg(d)
for i in x:
    print(i,': ',x[i])              
```

```python
def square_number(nums):
    for i in nums:
        yield(i*i)
       

my_nums  = square_number([1,2,3,4,5])
for num in my_nums:
    print(num)

my_nums2 = (x*x for x in [1,2,3,4,5])
print(list(my_nums2))
```

```python
def decorator_function(original_function):
    def wrapper_function(*args, **kwargs):
        print(f'wrapper executed this before the {original_function.__name__} function ')
        return original_function(*args, **kwargs)
    return wrapper_function    

@decorator_function
def display():
    print('dispay function ran')

@decorator_function
def display_info(name, age):
    print(f'display_info ran with {name}, {age}')   


display_info('John', 30)     


display()
```

##### Class Decorator

```python
class decorator_class(object):
    def __init__(self, original_function):
        self.original_function = original_function


    def __call__(self, *args, **kwargs):
        print(f'call executed this before the {self.original_function.__name__} function ')
        return self.original_function(*args, **kwargs)  



@decorator_class
def display():
    print('dispay function ran')

@decorator_class
def display_info(name, age):
    print(f'display_info ran with {name}, {age}')   


display_info('John', 30)     


display()    
    
```

```python
def deco(new_func):
    def wrap(*args, **kwargs):
        print('This is the wrapper function')
        return new_func(*args, **kwargs)
    return wrap

class deco_class(object):
    def __init__(self, new_func):
        self.new_func = new_func

    def __call__(self, *args, **kwargs):
        print('this is the call class')
        return self.new_func(*args, **kwargs)



@deco
@deco_class
def show_me():
    print('Both the wrapper function decorator and class decorator are on me')   

show_me()    
```

```python
from functools import wraps

def my_logger(orig_func):
    import logging
    logging.basicConfig(filename=f'{orig_func.__name__}.log', level=logging.INFO)

    @wraps(orig_func)
    def wrapper(*args, **kwargs):
        logging.info(f'Ran with args:{args}, and kwargs: {kwargs}')
        return orig_func(*args, **kwargs)
    return wrapper


def my_timer(orig_func):
    import time

    @wraps(orig_func)
    def wrapper(*args, **kwargs):
        t1 = time.time()
        result = orig_func(*args, **kwargs)
        t2 = time.time() - t1
        print(f'{orig_func.__name__}, ran in  {t2} sec')
        return result
    return wrapper

@my_logger
@my_timer
def display_info(name, age):
    print(f'display_info ran with {name}, {age}')   

display_info('Stephanie', 22)    

```

```python
import time

@my_timer
def display_info(name, age):
    time.sleep(2)
    print(f'display_info ran with argument ({name}, {age})')   

display_info('Victor', 22)
    
```

#### REDUCE

The reduce function is from function tools not from the global namespace like Map and Filter

```python
from functools import reduce

numbers = [1,2,3,4]

result = reduce(lambda x, y: x+y, numbers)
result2 = reduce(lambda x, y: x*y, numbers)

print(result)
print(result2)
```

#### Reduce function can have a starting point. See the example below.

```python
prices = [100, 200, 300, 400]

total = reduce(lambda total, price: total+price, prices, 1000)

print(total)
```

#### Example:

transactions = [5000, 12000, 8000, 20000, 15000]

The company wants to calculate the total transaction value.

But there's a complication:

Every transaction attracts a ₦500 processing fee.

So your job is to calculate the total amount including the ₦500 fee for every transaction.

```python
transactions = [5000, 12000, 8000, 20000, 15000]

all_transaction = list(map(lambda x: x+500, transactions))

total = reduce(lambda price, total: price+total, all_transaction)
print(total)
```

```python
transactions = [5000, 12000, 8000, 20000, 15000]

total = reduce(lambda price, total: price+total+500, transactions, 0)
print(total)
```
